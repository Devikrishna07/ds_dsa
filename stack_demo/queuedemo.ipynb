```python
class Order:
    def __init__(self,id,date,price,qty):
        self.id=id
        self.date=date
        self.price=price
        self.qty=qty
    def display(self):
        print(f"Orderid:{self.id} OrderDate:{self.date} OrderPrice:{self.price} OrderQty:{self.qty}")
              
```

```python
o1=Order(100,"10-10-2026",2200,2)
o2=Order(101,"15-07-2026",25000,3)
o3=Order(102,"04-04-2026",10000,4)
o4=Order(103,"04-10-2025",5000,2)
o5=Order(104,"10-09-2026",2500,2)
```

```python
o1.display()
```

```python
o2.display()
```

```python
o3.display()
```

```python
o4.display()
```

```python
from collections import deque
```

```python
order_queue=deque()
```

```python
order_queue.append(o1)
order_queue.append(o2)
order_queue.append(o3)
order_queue.append(o4)
order_queue.append(o5)
```

```python
while order_queue:
    cur=order_queue.popleft()
    print("processing order")
    cur.display()
```

```python
class Patient:
    def __init__(self,id,name,dept,type):
        self.id=id
        self.name=name
        self.dept=dept
        self.type=type
    def display(self):
        print(f"Patientid:{self.id} PatientName:{self.name} PatientDept:{self.dept} PatientType:{self.type}") 
       
```

```python
p1=Patient(101,"Amaya","Ortho","indoor")
p2=Patient(102,"Diya","ENT","indoor")
p3=Patient(103,"Sama","Physician","outdoor")
p4=Patient(104,"Lena","Pediac","outdoor")
p5=Patient(105,"Meenu","Cardiac","indoor")
```

```python
p1.display()
```

```python
p2.display()
```

```python
import heapq
```

```python
patients=[]
```

```python
heapq.heappush(patients,(3,p1))
heapq.heappush(patients,(1,p5))
heapq.heappush(patients,(2,p2))
heapq.heappush(patients,(5,p3))
heapq.heappush(patients,(4,p4))
```

```python
while patients:
    pr,p=heapq.heappop(patients)
    p.display()
```
