```python
class Student:
    def __init__(self,id,name):
        self.id=id
        self.name=name
    def std_info(self):
        print(f"Id:{self.id} name:{self.name}")
```

```python
s1=Student(1,"joe")
```

```python
s1.std_info()
```

```python
s1.id=100
```

```python
s1.std_info()
```

```python
class Customer:
    def __init__(self,id,name,address):
        self.id=id
        self.name=name
        self.address=address
    def cust_info(self):
        print(f"id{self.id} name{self.name} address{self.address}")
```

```python
c=Customer(100,"joe","goa")
```

```python
class Cart:
    def __init__(self,name,code,price,qty):
        self.name=name
        self.code=code
        self.price=price
        self.qty=qty
        self.total=price*qty
    def cart_info(self):
        print(f"name{self.name} code{self.code} price{self.price} qty{self.qty} total{self.total}")   
```

```python
c1=Cart("Laptop",100,50000,2)
```

```python
c1.cart_info()
```

```python
class Calculator:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def add(self):
        return self.a+self.b
    def sub(self):
        return self.a-self.b
    def mul(self):
        return self.a*self.b
    def divide(self):
        return self.a/self.b
```

```python
c=Calculator(10,30)
```

```python
c.add()
```

```python
c=Calculator(50,5)
```

```python
c.mul()
```

```python
c.divide()
```

```python
c.sub()
```
